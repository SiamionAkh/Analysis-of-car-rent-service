```python
df_sessions_clean = df_sessions_clean.drop(columns=['utm_keyword', 'device_os', 'device_brand', 'device_model','utm_campaign', 'utm_adcontent'])
```

### Preprocessing and Merging

```python
df_hits_clean.info(memory_usage='deep')
```

```python
df_hits_clean.head()
```

```python
df_hits_clean['hit_date'] = pd.to_datetime(df_hits_clean['hit_date'])
```

```python
columns_to_convert = ['hit_page_path', 'event_category', 'event_action']
for col in df_hits_clean.columns:
  if col in columns_to_convert:
    df_hits_clean[col] = df_hits_clean[col].astype('category')
```

```python
df_sessions_clean.info(memory_usage='deep')
```

```python
df_sessions_clean.head()
```

```python
df_sessions_clean['visit_datetime'] = pd.to_datetime(df_sessions_clean['visit_date'] + ' ' + df_sessions_clean['visit_time'])
```

```python
df_sessions_clean = df_sessions_clean.drop(columns=['visit_date', 'visit_time'])
```

```python
columns_to_convert = ['utm_source', 'utm_medium', 'device_category', 'device_screen_resolution', 'device_browser', 'geo_country', 'geo_city']
for col in df_sessions_clean.columns:
  if col in columns_to_convert:
    df_sessions_clean[col] = df_sessions_clean[col].astype('category')
```

```python
df_merged = pd.merge(df_sessions_clean, df_hits_clean, on='session_id', how='inner')
```

```python
del df_sessions_clean, df_hits_clean
```

## Analysis

###Analysis of traffic by cities

```python
target_events =  ['sub_car_claim_click', 'sub_car_claim_submit_click', 'sub_open_dialog_click', 'sub_custom_question_submit_click', 'sub_call_number_click', 'sub_callback_submit_click', 'sub_submit_success', 'sub_car_request_submit_click']
```

```python
df_merged['geo_city'].value_counts()
```

```python
main_cities = ['Moscow', 'Saint Petersburg']
```

```python
sessions_from_main_cities = len(df_merged[df_merged['geo_city'].isin(main_cities)])
sessions_from_other_cities = len(df_merged[~df_merged['geo_city'].isin(main_cities)])
```

```python
succ_sessions_from_main_cities = len(df_merged[(df_merged['geo_city'].isin(main_cities)) & (df_merged['event_action'].isin(target_events))])
succ_sessions_from_other_cities = len(df_merged[(~df_merged['geo_city'].isin(main_cities)) & (df_merged['event_action'].isin(target_events))])
#here and below, succ means success(success session contains target events)
```

```python
print(f'The ratio of successful sessions from main sities to all sessions from main cities {succ_sessions_from_main_cities / sessions_from_main_cities * 100}%')
```

```python
print(f'The ratio of successful sessions from other sities to all sessions from other cities {succ_sessions_from_main_cities / succ_sessions_from_other_cities * 100}%')
```

```python
print(f'The ratio of successful sessions from main sities to all successful sesions {succ_sessions_from_main_cities / (succ_sessions_from_other_cities + succ_sessions_from_main_cities) * 100}%')
```

```python
print(f'The ratio of sessions from main sities to all sesions {sessions_from_main_cities / (sessions_from_other_cities + sessions_from_main_cities) * 100}%')
```

```python
succ_sessions_per_city_count_df = df_merged[df_merged['event_action'].isin(target_events)].groupby('geo_city')['session_id'].count().reset_index()
succ_sessions_per_city_count_df.columns = ['geo_city', 'succ_session_count']
```

```python
all_sessions_per_city_count_df = df_merged.groupby('geo_city')['session_id'].count().reset_index()
all_sessions_per_city_count_df.columns = ['geo_city', 'all_sessions_in_city']
all_sessions_per_city_count_df = pd.merge(succ_sessions_per_city_count_df, all_sessions_per_city_count_df, on='geo_city', how='inner')
```

```python
all_sessions_per_city_count_df['succ_ratio'] = (all_sessions_per_city_count_df['succ_session_count'] / all_sessions_per_city_count_df['all_sessions_in_city']) * 100
```

```python
all_sessions_per_city_count_df = all_sessions_per_city_count_df.sort_values(by='succ_ratio', ascending=False)
display(all_sessions_per_city_count_df[all_sessions_per_city_count_df['all_sessions_in_city'] >= 100000].head(10))
```

##Conclusion

Analysis of traffic by cities shows that main source of traffic is 2 cities Moscow and Saint-Peterburg (58,4% and 63,2% counting only successful sessions)  
List of cities with the best CR (Conversation Rate) and over 100,000 sessions : ('Kazan', 'Grozny', 'Moscow', 'Krasnodar',	'Tula',	'Ufa',	'Samara', 'Balashikha',	'Rostov-on-Don',	'Saint Petersburg')

Analysis of traffic by device category shows that mobile devices are the most common for our users. At the same time, traffic from mobile devices has the best CR  
By the way tablet devices are very unpopular and have low CR

Analysis of traffic by sources and Analysis of traffic that comes from social media ad show that main sources are banner ads, paid search ads and traffic from other websites.  
Moreover ads in social media and Cost Per Action program provide a small amount of traffic, but with good CR

Analysis of car models successfulness shows that the best CR and the most in-demand are gasoline cars with automatic transmissions. Such as Hyundai Solaris, Toyota Camry 7 (XV50), Volkswagen Polo 6

Analisis of organic and paid (nonorganic) traffic shows that paid traffic prevail in amount (63% of all traffic) and is better in CR (0.71% vs 0.62% from organic traffic)
